```python
import os
```

```python
# Parameters 
sub_name = 'S10'
task_name = 'auditory'

# Set the path to the directory containing the data
raw_folder = '/Users/joaopereira/Downloads/2024-02-21_009/2024-02-21_010'
raw_folder = '/Volumes/T7/data/RAW_FNIRS_music'

# Define output folder
out_folder = os.path.join('..', 'data', 'dataset', sub_name, 'nirs')

# if folder does not exist, create it
if not os.path.exists(out_folder):
    os.makedirs(out_folder)
```

```python
# Rename files in a directory following the BIDS format
# https://bids.neuroimaging.io/


# Get all files in the directory
files = os.listdir(raw_folder)

# Loop through all files

for file in files:

    print(file)

    # Split the file name to get the different parts
    parts = file.split('_')

    print(parts)


    # Check if file is a file.
    if not file.startswith('.')  and os.path.isfile(os.path.join(raw_folder, file)):
        
        # If the file has the correct number of parts
        if len(parts) == 3:
            # Get the different parts
            date = parts[0]

            run = parts[1]

            type = parts[2].split('.')[0]

            # Get file extension.
            extension = parts[2].split('.')[1]

            # Create the new name
            new_name = f'sub-{sub_name}_task-{task_name}_run-{run}_{type}.{extension}'

            # Rename the file
            os.rename(os.path.join(raw_folder, file), os.path.join(out_folder, new_name))
            print(f'{file} -> {new_name}')
        elif len(parts) == 2:

            # Get the different parts
            date = parts[0]
            run = parts[1].split('.')[0]

            # Get file extension.
            extension = parts[1].split('.')[1]


            # Change the name of the file
            new_name = f'sub-{sub_name}_task-{task_name}_run-{run}.{extension}'

            os.rename(os.path.join(raw_folder, file), os.path.join(out_folder, new_name))
        else:
            # Just copy the file.
            os.rename(os.path.join(raw_folder, file), os.path.join(out_folder, file))


        

```
